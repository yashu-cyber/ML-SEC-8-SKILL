```python
# ============================================================
# EXPERIMENT 5
# LOGISTIC REGRESSION - L1, L2 AND ELASTIC NET
# DATASET: AGE INSURANCE
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv("/content/age_insurance (1).csv")

print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
print(df.head())

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 3. TARGET VARIABLE
# ============================================================

# Insurance:
# 0 = Not Insured
# 1 = Insured

y = df["Insurance"]


# ============================================================
# 4. SELECT PREDICTOR VARIABLES
# ============================================================

features = [
    "Age",
    "InsurancePlan"
]

X = df[features].copy()


# ============================================================
# 5. CATEGORICAL VARIABLES
# ============================================================

categorical_features = [
    "InsurancePlan"
]


# ============================================================
# 6. NUMERICAL VARIABLES
# ============================================================

numerical_features = [
    "Age"
]


# ============================================================
# 7. HANDLE MISSING VALUES
# ============================================================

for col in numerical_features:
    X[col] = X[col].fillna(X[col].median())

for col in categorical_features:
    X[col] = X[col].fillna(X[col].mode()[0])


# ============================================================
# 8. TRAIN-TEST SPLIT
# ============================================================

# stratify=y keeps the same proportion of 0 and 1
# in both training and testing data.

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ============================================================
# 9. PREPROCESSING
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),

        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            ),
            categorical_features
        )
    ]
)


# ============================================================
# 10. NORMAL BINOMIAL LOGISTIC REGRESSION
# ============================================================

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        (
            "classifier",
            LogisticRegression(
                max_iter=2000,
                random_state=42
            )
        )
    ]
)


# ============================================================
# 11. TRAIN MODEL
# ============================================================

logistic_model.fit(X_train, y_train)


# ============================================================
# 12. PREDICTION
# ============================================================

y_pred = logistic_model.predict(X_test)

y_probability = logistic_model.predict_proba(X_test)[:, 1]


# ============================================================
# 13. MODEL EVALUATION
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

print("\n============================================")
print("BINOMIAL LOGISTIC REGRESSION RESULTS")
print("============================================")

print("\nAccuracy:")
print(round(accuracy, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Not Insured", "Insured"]
    )
)

print("\nROC-AUC:")
print(round(roc_auc_score(y_test, y_probability), 4))


# ============================================================
# 14. L1 LOGISTIC REGRESSION
# ============================================================

l1_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        (
            "classifier",
            LogisticRegression(
                penalty="l1",
                solver="liblinear",
                C=1.0,
                max_iter=2000,
                random_state=42
            )
        )
    ]
)


# ============================================================
# 15. TRAIN L1 MODEL
# ============================================================

l1_model.fit(X_train, y_train)


# ============================================================
# 16. L1 PREDICTION
# ============================================================

y_pred_l1 = l1_model.predict(X_test)

y_prob_l1 = l1_model.predict_proba(X_test)[:, 1]


# ============================================================
# 17. L1 MODEL EVALUATION
# ============================================================

accuracy_l1 = accuracy_score(y_test, y_pred_l1)

print("\n============================================")
print("L1 LOGISTIC REGRESSION RESULTS")
print("============================================")

print("\nAccuracy:")
print(round(accuracy_l1, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_l1))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_l1,
        target_names=["Not Insured", "Insured"]
    )
)

print("\nROC-AUC:")
print(round(roc_auc_score(y_test, y_prob_l1), 4))


# ============================================================
# 18. L2 LOGISTIC REGRESSION
# ============================================================

l2_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        (
            "classifier",
            LogisticRegression(
                penalty="l2",
                solver="lbfgs",
                C=1.0,
                max_iter=2000,
                random_state=42
            )
        )
    ]
)


# ============================================================
# 19. TRAIN L2 MODEL
# ============================================================

l2_model.fit(X_train, y_train)


# ============================================================
# 20. L2 PREDICTION
# ============================================================

y_pred_l2 = l2_model.predict(X_test)

y_prob_l2 = l2_model.predict_proba(X_test)[:, 1]


# ============================================================
# 21. L2 MODEL EVALUATION
# ============================================================

accuracy_l2 = accuracy_score(y_test, y_pred_l2)

print("\n============================================")
print("L2 LOGISTIC REGRESSION RESULTS")
print("============================================")

print("\nAccuracy:")
print(round(accuracy_l2, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_l2))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_l2,
        target_names=["Not Insured", "Insured"]
    )
)

print("\nROC-AUC:")
print(round(roc_auc_score(y_test, y_prob_l2), 4))


# ============================================================
# 22. ELASTIC NET LOGISTIC REGRESSION
# ============================================================

elastic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        (
            "classifier",
            LogisticRegression(
                penalty="elasticnet",
                solver="saga",
                l1_ratio=0.5,
                C=1.0,
                max_iter=5000,
                random_state=42
            )
        )
    ]
)


# ============================================================
# 23. TRAIN ELASTIC NET MODEL
# ============================================================

elastic_model.fit(X_train, y_train)


# ============================================================
# 24. ELASTIC NET PREDICTION
# ============================================================

y_pred_elastic = elastic_model.predict(X_test)

y_prob_elastic = elastic_model.predict_proba(X_test)[:, 1]


# ============================================================
# 25. ELASTIC NET MODEL EVALUATION
# ============================================================

accuracy_elastic = accuracy_score(
    y_test,
    y_pred_elastic
)

print("\n============================================")
print("ELASTIC NET LOGISTIC REGRESSION RESULTS")
print("============================================")

print("\nAccuracy:")
print(round(accuracy_elastic, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_elastic))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_elastic,
        target_names=["Not Insured", "Insured"]
    )
)

print("\nROC-AUC:")
print(
    round(
        roc_auc_score(
            y_test,
            y_prob_elastic
        ),
        4
    )
)


# ============================================================
# 26. COMPARE ALL MODELS
# ============================================================

print("\n============================================")
print("MODEL COMPARISON")
print("============================================")

print("Normal Logistic Regression Accuracy:",
      round(accuracy, 4))

print("L1 Logistic Regression Accuracy:",
      round(accuracy_l1, 4))

print("L2 Logistic Regression Accuracy:",
      round(accuracy_l2, 4))

print("Elastic Net Accuracy:",
      round(accuracy_elastic, 4))


# ============================================================
# 27. PREDICT FOR A NEW PERSON
# ============================================================

new_person = pd.DataFrame([
    {
        "Age": 50,
        "InsurancePlan": 2
    }
])


# ============================================================
# 28. PREDICTION USING ELASTIC NET
# ============================================================

prediction = elastic_model.predict(new_person)[0]

probability = elastic_model.predict_proba(
    new_person
)[0, 1]


# ============================================================
# 29. DISPLAY RESULT
# ============================================================

print("\n============================================")
print("NEW PERSON INSURANCE PREDICTION")
print("============================================")

if prediction == 1:
    print("Predicted Insurance Status : INSURED")
else:
    print("Predicted Insurance Status : NOT INSURED")

print(
    f"Probability of Insurance   : {probability * 100:.2f}%"
)

print(
    f"Probability of Not Insured : {(1 - probability) * 100:.2f}%"
)
```
