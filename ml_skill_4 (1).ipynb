```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/colab/placement_dataset.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig1 = plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Branch")

plt.title("Placement Count by Branch")
plt.xlabel("Branch")
plt.ylabel("Count")

plt.show()
```

```python
fig2 = plt.figure(figsize=(6,4))

plt.hist(df["CGPA"].dropna(),
         bins=15,
         color="skyblue",
         edgecolor="black")

plt.title("CGPA Distribution")
plt.xlabel("CGPA")
plt.ylabel("Frequency")

plt.show()
```

```python
fig3 = plt.figure(figsize=(6,5))

numeric_df = df.select_dtypes(include="number")

corr_matrix = numeric_df.corr()

sns.heatmap(corr_matrix,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")

plt.show()
```

```python
from matplotlib.backends.backend_pdf import PdfPages

# Define the output PDF file path in Google Drive
pdf_path = '/content/drive/MyDrive/colab/all_plots.pdf'

with PdfPages(pdf_path) as pdf:
    pdf.savefig(fig1)
    pdf.savefig(fig2)
    pdf.savefig(fig3)

print(f"All plots saved to {pdf_path}")
```
