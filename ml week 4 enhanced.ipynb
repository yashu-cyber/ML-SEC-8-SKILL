```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset directly from the local environment
df = pd.read_csv('placement_dataset.csv')

# 1st Graph: Placement Count by Branch
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Branch")
plt.title("Placement Count by Branch")
plt.xlabel("Branch")
plt.ylabel("Count")
plt.show()

# 2nd Graph: CGPA Distribution
plt.figure(figsize=(6,4))
plt.hist(df["CGPA"].dropna(),
         bins=15,
         color="skyblue",
         edgecolor="black")
plt.title("CGPA Distribution")
plt.xlabel("CGPA")
plt.ylabel("Frequency")
plt.show()

# 3rd Graph: Correlation Heatmap
plt.figure(figsize=(6,5))
numeric_df = df.select_dtypes(include="number")
corr_matrix = numeric_df.corr()
sns.heatmap(corr_matrix,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Identify the placement column
placement_col = df.columns[-1]

# Create a visually appealing Violin Plot
plt.figure(figsize=(10, 6))
# Fixed the warning by assigning x to hue and setting legend=False
sns.violinplot(data=df, x=placement_col, y='CGPA', hue=placement_col, palette='Pastel1', inner='quartile', legend=False)

# Adding aesthetic details
plt.title(f'Distribution of CGPA by {placement_col}', fontsize=16, pad=20)
plt.xlabel(placement_col, fontsize=12)
plt.ylabel('CGPA', fontsize=12)
sns.despine()

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Set the style for a cleaner look
sns.set_theme(style='whitegrid')

# Create the Bubble Chart
plt.figure(figsize=(12, 8))

# Aspect 1: CGPA (X), Aspect 2: IQ (Y), Aspect 3: Aptitude_Score (Size)
bubble_plot = sns.scatterplot(
    data=df.dropna(subset=['CGPA', 'IQ', 'Aptitude_Score']),
    x='CGPA',
    y='IQ',
    size='Aptitude_Score',
    hue='Branch',       # Extra dimension for visual appeal
    sizes=(20, 500),    # Control the range of bubble sizes
    alpha=0.6,          # Transparency makes overlapping bubbles easier to see
    palette='viridis'
)

# Move the legend outside the plot for better visibility
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)

# Add titles and labels
plt.title('3-Way Analysis: CGPA vs. IQ (Size = Aptitude Score)', fontsize=16)
plt.xlabel('CGPA', fontsize=12)
plt.ylabel('IQ', fontsize=12)

plt.tight_layout()
plt.show()
```
